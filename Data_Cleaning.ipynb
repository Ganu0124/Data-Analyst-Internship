```python
median_income = df["Income"].median()

print("Median income:", median_income)
```

```python
df["Income"] = df["Income"].fillna(median_income)

print("Missing income:", df["Income"].isnull().sum())
```

```python
print("Before:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("After:", df.duplicated().sum())
```

```python
print("Remaining missing values:")
print(df.isnull().sum())

print("\nRemaining duplicates:")
print(df.duplicated().sum())

print("\nDataset shape:")
print(df.shape)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(df.columns.tolist())
```

```python
df.select_dtypes(include="object").head()
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["education"] = df["education"].str.title()

print(df["education"].unique())
```

```python
df["dt_customer"] = pd.to_datetime(
    df["dt_customer"],
    dayfirst=True,
    errors="coerce"
)

print(df["dt_customer"].dtype)
print(df["dt_customer"].head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nEducation values:")
print(df["education"].unique())

print("\nDate data type:")
print(df["dt_customer"].dtype)
```

```python
print(df.dtypes)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numeric columns:")
print(numeric_columns.tolist())
```

```python
df.describe()
```

```python
Q1 = df[numeric_columns].quantile(0.25)
Q3 = df[numeric_columns].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = (
    (df[numeric_columns] < lower_limit) |
    (df[numeric_columns] > upper_limit)
)

print("Outliers in each column:")
print(outliers.sum())
```

```python
print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Invalid dates:", df["dt_customer"].isnull().sum())
```

```python
df.to_csv("cleaned_marketing_campaign.csv",index=False)
print("Cleaned dataset exported successfully!")
```
